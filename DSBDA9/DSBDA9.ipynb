```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/PythonXYZ/DSBDA/DSBDA8/Titanic-Dataset.csv")
```

```python
df
```

```python
# Select relevant columns with correct capitalization
df = df[['Sex', 'Age', 'Survived']]
```

```python
# Drop rows with missing Age values
df = df.dropna(subset=['Age'])
```

```python
# Plot 1: Boxplot of Age vs. Sex
plt.figure(figsize=(8, 6))
sns.boxplot(x='Sex', y='Age', data=df)
plt.title('Age Distribution by Gender')
plt.show()
```

```python
# Plot 2: Boxplot of Age vs. Sex with Survival info
plt.figure(figsize=(8, 6))
sns.boxplot(x='Sex', y='Age', hue='Survived', data=df)
plt.title('Age Distribution by Gender and Survival')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```
