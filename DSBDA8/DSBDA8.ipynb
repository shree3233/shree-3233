```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/PythonXYZ/DSBDA/DSBDA8/Titanic-Dataset.csv")
```

```python
# Show basic info and first few rows
print(df.info())
```

```python
print(df.head())
```

```python
sns.countplot(x='Survived', hue='Sex', data=df)
plt.title('Survival Count by Sex')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.legend(title='Sex')
plt.show()
```

```python
sns.countplot(x='Survived', hue='Pclass', data=df)
plt.title('Survival Count by Passenger Class')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.legend(title='Pclass')
plt.show()
```

```python
sns.histplot(data=df, x='Age', hue='Survived', kde=True, bins=30)
plt.title('Age Distribution by Survival')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived')
plt.show()
```

```python
# 4. Hist Plot: Fare distribution
sns.histplot(df['Fare'], bins=40, kde=True)
plt.title('Distribution of Ticket Fare')
plt.xlabel('Fare')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
# 5. KDE Plot: Age distribution with survival
sns.kdeplot(data=df, x='Age', hue='Survived', fill=True)
plt.title('KDE of Age Distribution by Survival')
plt.xlabel('Age')
plt.ylabel('Density')
plt.legend(title='Survived')
plt.show()
```

```python
# 6. Bar Plot: Average fare by passenger class
sns.barplot(x='Pclass', y='Fare', data=df, ci=None)
plt.title('Average Fare by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Average Fare')
plt.show()
```

```python
# 7. Count Plot: Survival by embarkation location
sns.countplot(x='Survived', hue='Embarked', data=df)
plt.title('Survival Count by Embarked Location')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.legend(title='Embarked')
plt.show()
```

```python
sns.pairplot(df[['Survived', 'Age', 'Fare', 'Pclass']])
plt.suptitle('Pairplot of Survival, Age, Fare, and Passenger Class', y=1.02)
plt.show()
```

```python
sns.jointplot(x='Age', y='Fare', data=df, kind='scatter')
plt.suptitle('Jointplot of Age vs Fare', y=1.02)
plt.show()
```
